````python
#Move 2x2 jumping 2 pixel time and compute average.

def resize(myimage):
    new_image = np.zeros((myimage.shape[0]//2, myimage.shape[1]//2))
    for i in range(new_image.shape[0]):
        for j in range(new_image.shape[1]):
            total = 0
            total += myimage[2*i][2*j]
            total += myimage[2*i+1][2*j]
            total += myimage[2*i][2*j+1]
            total += myimage[2*i+1][2*j+1]
            new_image[i][j] = total//4
    return new_image
````

````python
#Move 2x2 jumping 2 pixel time and compute average.

def resize1(myimage, k=2):
    new_image = np.zeros((myimage.shape[0]//k, myimage.shape[1]//k))
    for i in range(new_image.shape[0]):
        for j in range(new_image.shape[1]):
            
            total = 0
            ii = k*i
            jj = k*j
            for ki in range(k):
                for kj in range(k):
                    total += myimage[ii+ki][jj+kj]
            new_image[i][j] = total//(k*k)
    return new_image
````
